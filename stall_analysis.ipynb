```python
# Gikomba Market Stall Performance Analysis

**Author:** Harith Parvez · PLP Data Analytics Cohort 7
**Objective:** Analyze 14 days of sales and current stock levels for Wanjiku's fresh-produce stall to evaluate business growth, identify peak/slow periods, and provide inventory restock guidance.
```

```python
# Task: Load the 14-day sales records and inventory dictionary
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700]   # 14 days, KES
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2

stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions":   {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma":   {"qty": 9,  "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado":  {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}
print("Dataset loaded successfully.")
```

```python
# Task 1: Compute total, average, highest, and lowest sales, plus day/week lookup
total_sales = sum(daily_sales)
avg_sales = total_sales / len(daily_sales)
max_sales = max(daily_sales)
min_sales = min(daily_sales)

best_pos = daily_sales.index(max_sales)
worst_pos = daily_sales.index(min_sales)

best_day_name = days[best_pos]
best_week = (best_pos // 7) + 1

worst_day_name = days[worst_pos]
worst_week = (worst_pos // 7) + 1

print(f"Total 14-Day Sales: KES {total_sales:,}")
print(f"Average Daily Sales: KES {avg_sales:,.2f}")
print(f"Best Day: {best_day_name} (Week {best_week}) with KES {max_sales:,}")
print(f"Worst Day: {worst_day_name} (Week {worst_week}) with KES {min_sales:,}")
```

```python
# Task 2: Calculate week totals via slicing and determine growth percentage
def week_totals(sales):
    """Return a tuple containing Week 1 total and Week 2 total sales."""
    return sum(sales[:7]), sum(sales[7:])

w1_total, w2_total = week_totals(daily_sales)
wow_growth = ((w2_total - w1_total) / w1_total) * 100

print(f"Week 1 Total: KES {w1_total:,}")
print(f"Week 2 Total: KES {w2_total:,}")
print(f"Week-on-Week Growth: {wow_growth:.1f}%")
```

```python
# Task 3: Classify each day relative to average and count categories
def classify_day(amount, average):
    """Classify day sales as Strong (>20% above avg), Weak (>20% below avg), or Normal."""
    if amount > average * 1.2:
        return "Strong"
    elif amount < average * 0.8:
        return "Weak"
    else:
        return "Normal"

category_counts = {}
for sale in daily_sales:
    cat = classify_day(sale, avg_sales)
    category_counts[cat] = category_counts.get(cat, 0) + 1

print("Day Classifications Count:")
for category, count in category_counts.items():
    print(f"- {category}: {count} days")
```

```python
# Task 4: Compare weekday (Mon-Fri) and weekend (Sat-Sun) performance
weekday_sales = [sale for d, sale in zip(days, daily_sales) if d not in ["Sat", "Sun"]]
weekend_sales = [sale for d, sale in zip(days, daily_sales) if d in ["Sat", "Sun"]]

avg_weekday = sum(weekday_sales) / len(weekday_sales)
avg_weekend = sum(weekend_sales) / len(weekend_sales)
weekend_premium = ((avg_weekend - avg_weekday) / avg_weekday) * 100

print(f"Average Weekday Sales: KES {avg_weekday:,.2f}")
print(f"Average Weekend Sales: KES {avg_weekend:,.2f}")
print(f"Weekend Daily Average is {weekend_premium:.1f}% higher than Weekdays.")
```

```python
# Task 5: Identify items to restock and calculate gross margins per product
def check_reorder(inventory):
    """Return a list of product names that are at or below their reorder level."""
    restock_list = []
    for product, info in inventory.items():
        if info["qty"] <= info["reorder"]:
            restock_list.append(product)
    return restock_list

items_to_restock = check_reorder(stock)

print("Product Margins & Reorder Status:")
for product, info in stock.items():
    margin_pct = ((info["price"] - info["cost"]) / info["price"]) * 100
    status = "REORDER" if product in items_to_restock else "OK"
    print(f"- {product:<10} | Margin: {margin_pct:>5.1f}% | Stock: {info['qty']}kg (Reorder: {info['reorder']}kg) -> {status}")
```

```python
# Task 6: Print clean, professional stall performance report
print("=" * 45)
print("       STALL PERFORMANCE REPORT - GIKOMBA")
print("=" * 45)
print(f"Total 2-Week Revenue : KES {total_sales:,}")
print(f"Week 1 Revenue       : KES {w1_total:,}")
print(f"Week 2 Revenue       : KES {w2_total:,}")
print(f"Week-on-Week Growth  : {wow_growth:.1f}%")
print(f"Average Daily Sales  : KES {avg_sales:,.2f}")
print(f"Peak Sales Day       : {best_day_name} (KES {max_sales:,})")
print(f"Slowest Sales Day    : {worst_day_name} (KES {min_sales:,})")
print("-" * 45)
print(f"Items Needing Restock: {', '.join(items_to_restock)}")
print("=" * 45)
```

```python
## Recommendations for Wanjiku

1. **Capitalize on Weekend Traffic:** Weekends generate significantly higher revenue, with Saturday hitting peak sales of KES 5,600. Wanjiku should ensure full stock availability and extended operating hours ahead of Fridays and Saturdays.
2. **Address Mid-Week Slumps:** Weekdays average lower sales figures (with Tuesday dropping to a low of KES 1,950). Introducing targeted weekday promotions or bundle discounts on slow days can help level out cash flow.
3. **Urgent Restock Action:** Inventory monitoring shows that **Tomatoes** (18kg remaining vs. 25kg reorder threshold) and **Sukuma** (9kg remaining vs. 15kg reorder threshold) require immediate replenishment to prevent stock-outs.
4. **Focus on High-Margin Produce:** Avocado yields the highest gross margin at 42.9%, making it a prime candidate for prominent stall placement and upselling.
```
